```python
import pandas as pd

# 1. Ensure the Date column is in datetime format
df["Date"] = pd.to_datetime(df["Date"])

# 2. Find the latest Date in the entire dataset (to calculate Recency - R)
latest_Date = df["Date"].max()

# 3. Perform the groupby and aggregation
rfm_df = (
    df.groupby("CustomerKey")
    .agg(
        # F (Frequency): Count unique invoice numbers for each customer
        F=("InvoNo", "nunique"),
        # M (Monetary): Sum of the amount for each customer
        M=("amount", "sum"),
        # Temporary step to find the latest transaction Date per customer for R
        last_trans_Date=("Date", "max"),
    )
    .reset_index()
)

# 4. Calculate R (Recency): Difference between the dataset's latest Date and customer's latest transaction Date
rfm_df["R"] = (latest_Date - rfm_df["last_trans_Date"]).dt.days

# 5. Rename 'CustomerKey' to 'customer' and drop the temporary Date column
rfm_df = rfm_df.rename(columns={"CustomerKey": "customer"})
rfm_df = rfm_df[["customer", "F", "M", "R"]]  # Reorder columns as requested

# View the result
print(rfm_df)
```
